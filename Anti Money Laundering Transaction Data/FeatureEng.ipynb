```python
import os, sys, platform, yaml, re
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import DoubleType, FloatType, StringType
from pyspark.sql.dataframe import DataFrame
import pandas as pd
from py2neo import Graph
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
conf_path = str(Path.cwd() / "config" / "ETL_config.yaml")
with open(conf_path, "r") as f:
    CFG = yaml.safe_load(f)
    
NEO4J_URI  = CFG["neo4j"]["uri"]          # "bolt://localhost:7687"
NEO4J_USER = CFG["neo4j"]["user"]
NEO4J_PASS = CFG["neo4j"]["pass"]
NEO4J_DB   = CFG["neo4j"]["database"]

opts = {
    "url": NEO4J_URI,
    "authentication.type": "basic",
    "authentication.basic.username": NEO4J_USER,
    "authentication.basic.password": NEO4J_PASS,
    "database": NEO4J_DB,
}
```

```python
graph = Graph(NEO4J_URI, auth=(NEO4J_USER, NEO4J_PASS), name=NEO4J_DB)
```

# Features temporales

```python
query = """
MATCH (n)
UNWIND keys(n) AS k
RETURN k AS property_key, count(*) AS nodes_with_key
ORDER BY nodes_with_key DESC, property_key
"""

df_props = graph.run(query).to_data_frame()
df_props.head(20)
```

```python
query = """
MATCH ()-[r]->()
UNWIND keys(r) AS k
RETURN k AS property_key, count(*) AS rels_with_key
ORDER BY rels_with_key DESC, property_key
"""
df_rel_props = graph.run(query).to_data_frame()
df_rel_props.head(20)
```
